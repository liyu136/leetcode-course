### 二、从思路到代码

练习3：  
状态： 
初始化：次数count = 0   
更新条件：循环中出现了数字=target   
更新方式：count+1  
返回值：count  

```python
def count_target(nums, target):
    """返回 target 在 nums 中出现的次数。"""
    if nums is None:
        return 0
    count = 0
    for x in nums:
        if x == target:
            count +=1
    return count
```

```python
assert count_target([1, 2, 2, 3], 2) == 2
assert count_target([], 2) == 0
assert count_target([1, 3], 2) == 0
```

```python
# 练习4
def find_min(nums):
    """返回非空数组中的最小值。"""
    min = nums[0]
    for x in nums:
        if x < min:
            min = x
    return min

# 时间复杂度 o(n)
# 空间复杂度 o(1) 因为只建立了两个变量来存放数据
```

订正：不要用min作为变量名，因为min是自带的函数

```python
# 练习 5：判断重复——暴力版本
def contains_duplicate_brute_force(nums):

    if nums is None:
        return False

    for ii in range(len(nums)-1): #从0开始到倒数第二个数
        for jj in range(ii+1 , len(nums)): #从ii的后一个index开始到最后一个数
            if nums[ii] == nums[jj]: #发现重复
                return True
    
    return False

# 时间复杂度 o(n^2)，因为使用了双重循环
# 空间复杂度 o(1)，没有建立什么新的长的储存变量
```

```python
# 练习6：集合版本
# 集合中保存见过的元素
# 先检查、后加入

def contains_duplicate(nums):
    if nums is None:
        return False
        
    seen=set()
    seen.add(nums[0]) #见过第一个数字了

    for x in nums[1:]: #从第二个元素开始
        if x in seen:
            return True
        else:
            seen.add(x)

    return False
# 最坏情况下集合会保存len(nums)个元素
```

```python
contains_duplicate([])
```

订正： 如果普通循环本身就能覆盖边界情况，尽量不要单独处理第一个元素。

```python
def contains_duplicate(nums):
        
    seen=set()

    for x in nums: 
        if x in seen:
            return True
        else:
            seen.add(x)

    return False
```

### 四、综合练习   
练习 12：两数之和暴力版   
外层下标ii范围：0到len(nums)-2  
内层下标jj范围：从 ii + 1 开始到 len(nums)-1  
满足条件的布尔表达式：nums[ii]+num[jj] == target  
返回内容：return [ii,jj]  
时间复杂度：o(n^2)  
额外空间复杂度：o(1)  

```python
def two_sum_brute_force(nums, target):
    for ii in range(len(nums)-1):
        for jj in range(ii+1,len(nums)):
            if nums[ii]+nums[jj] == target:
                return [ii,jj]
    
```

```python
assert two_sum_brute_force([2, 7, 11, 15], 9) == [0, 1]
assert two_sum_brute_force([3, 2, 4], 6) == [1, 2]
assert two_sum_brute_force([3, 3], 6) == [0, 1]
```
