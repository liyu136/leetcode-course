## 2. 原地反转

不使用 `reverse()`、切片反转或新列表，原地反转 `nums`。函数不需要返回值。

编码前写出：两个下标的初始值、循环条件、每轮更新，以及循环不变量。

原地反转   
初始left=0，right=n-1，分别从头和尾开始  
循环条件：left < right，如果是偶数，正好平分，如果是奇数，最中间的那个会不动     
每轮更新：left+=1，right-=1    


````python
def reverse_in_place(nums):
    left = 0
    right = len(nums)-1
    
    while left < right:
        nums[left], nums[right] = nums[right], nums[left]
        left+=1
        right-=1
````

````python
a = []
reverse_in_place(a)
assert a == []

a = [1]
reverse_in_place(a)
assert a == [1]

a = [1, 2, 3, 4, 5]
result = reverse_in_place(a)
assert a == [5, 4, 3, 2, 1]
assert result is None
````

## 3. 删除指定值

把所有不等于 `val` 的元素稳定地写到 `nums` 前部，返回有效长度 `k`。只检查 `nums[:k]`，后面的内容不作要求。不能创建与输入规模相关的新列表，也不要调用 `remove`。

编码前写出 `read` 和 `write` 各自表示什么，以及 `nums[:write]` 始终满足什么性质。

read是读到的数值对应的指针，write是写入的数值对应的指针，



````python
def remove_value(nums, val):
    write = 0

    for read in range(len(nums)):
        if nums[read] != val:
            nums[write] = nums[read]
            write+=1

    return write
````

````python
a = [3, 2, 2, 3, 4]
k = remove_value(a, 3)
assert k == 3
assert a[:k] == [2, 2, 4]

a = []
assert remove_value(a, 1) == 0

a = [1, 1]
assert remove_value(a, 1) == 0

a = [1, 2]
k = remove_value(a, 3)
assert k == 2 and a[:k] == [1, 2]
````

## 4. 移动零

把所有 `0` 移动到列表末尾，同时保持非零元素的相对顺序。必须原地修改，不能创建长度随 `n` 增长的新列表，函数返回 `None`。

示例：`[0, 1, 0, 3, 12] → [1, 3, 12, 0, 0]`。

提示：先思考它与 `remove_value` 的相同点和不同点。不要直接删除元素。

````python
def move_zeroes(nums):
    write = 0
    for read in range(len(nums)):
        if nums[read] != 0:
            nums[write] = nums[read]
            write+=1
    nums[write:] = [0]*(len(nums)-write)
    return None
````

````python
cases = [
    ([0, 1, 0, 3, 12], [1, 3, 12, 0, 0]),
    ([], []),
    ([0], [0]),
    ([1, 2], [1, 2]),
    ([0, 0, 1], [1, 0, 0]),
]

for original, expected in cases:
    a = original.copy()
    result = move_zeroes(a)
    assert a == expected, (original, a, expected)
    assert result is None
````

## 5. 挑战：有序数组的平方

给定一个非递减整数数组，返回每个元素平方后仍按非递减顺序排列的新列表。要求 `O(n)` 时间。

示例：`[-4, -1, 0, 3, 10] → [0, 1, 9, 16, 100]`。

不要调用 `sort()` 或 `sorted()`。先回答：最大的平方可能来自哪里？结果列表应该从小到大填，还是从大到小填？

````python
def sorted_squares(nums):
    if nums == []:
        return []
    left = 0
    right = len(nums)-1
    new_nums = nums.copy()
    write = len(nums)-1

    while left < right:
        if nums[left]**2 < nums[right]**2:
            new_nums[write] = nums[right]**2
            right -= 1
        else:
            new_nums[write] = nums[left]**2
            left+=1
        write-=1
    new_nums[write] = nums[left]**2
    return new_nums
````

````python
assert sorted_squares([]) == []
assert sorted_squares([-4, -1, 0, 3, 10]) == [0, 1, 9, 16, 100]
assert sorted_squares([-7, -3, 2, 3, 11]) == [4, 9, 9, 49, 121]
assert sorted_squares([-3, -2, -1]) == [1, 4, 9]
assert sorted_squares([1, 2, 3]) == [1, 4, 9]
````

## 6. 数组右旋

把列表向右旋转 `k` 步，必须让调用者观察到原列表发生改变。允许先使用 `O(n)` 额外空间。

示例：`[1,2,3,4,5,6,7]` 右旋 3 步得到 `[5,6,7,1,2,3,4]`。

注意处理空列表、`k=0` 和 `k` 大于列表长度。完成基础版本后，可选挑战：改为 `O(1)` 额外空间。

````python
def rotate_right(nums, k):
    # O(N)
    if nums == []:
        return None

    new = [0]*len(nums)
    for i in range(len(nums)):
        j = (i+k) % len(nums)
        new[j]=nums[i]
    nums[:] = new

    return None
````

````python
def rotate_right(nums, k):
    # O(1)
    if nums == []:
        return None

    def reverse1(nums,left,right):
        # nums,start,end
        while left < right:
            nums[left],nums[right] = nums[right],nums[left]
            left+=1
            right-=1
    
    reverse1(nums,0,len(nums)-1)
    k = k % len(nums)
    reverse1(nums,0,k-1)
    reverse1(nums,k,len(nums)-1)
````

````python
cases = [
    ([1, 2, 3, 4, 5, 6, 7], 3, [5, 6, 7, 1, 2, 3, 4]),
    ([], 4, []),
    ([1], 10, [1]),
    ([1, 2, 3], 0, [1, 2, 3]),
    ([1, 2, 3], 4, [3, 1, 2]),
]

for original, k, expected in cases:
    a = original.copy()
    result = rotate_right(a, k)
    assert a == expected, (original, k, a, expected)
    assert result is None
````
