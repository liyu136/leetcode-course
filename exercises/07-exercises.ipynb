每道递归题编码前先写函数契约：输入节点代表哪棵树，返回值表示什么，空树返回什么。

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def sample_tree():
    #       1
    #      / \
    #     2   3
    #    / \
    #   4   5
    return TreeNode(1, TreeNode(2, TreeNode(4), TreeNode(5)), TreeNode(3))
```

在这里作答。

1.前序：立即处理当前节点。中序：处理完left后。后续：left和right都处理完之后

## 2. 三种 DFS 遍历

分别实现前序、中序、后序遍历，返回节点值列表。使用递归。

```python
def preorder(root):
    result = []

    if root is None:
        return []

    def preorder_helper(node):
        if node is None:
            return
        result.append(node.val)
        preorder_helper(node.left)
        preorder_helper(node.right)

    preorder_helper(root)
    return result

def inorder(root):
    result = []
    if root is None:
        return []
    def inorder_helper(node):
        if node is None:
            return
        inorder_helper(node.left)
        result.append(node.val)
        inorder_helper(node.right)
    inorder_helper(root)
    return result

def postorder(root):
    result = []
    if root is None:
        return []
    def postorder_helper(node):
        if node is None:
            return
        postorder_helper(node.left)
        postorder_helper(node.right)
        result.append(node.val)
    postorder_helper(root)
    return result   
    return result
```

```python
root = sample_tree()
assert preorder(None) == []
assert preorder(root) == [1,2,4,5,3]
assert inorder(root) == [4,2,5,1,3]
assert postorder(root) == [4,5,2,3,1]
```

## 3. 最大深度

定义空树深度为 0，单节点树深度为 1。不要使用全局列表。写出递推关系。

契约：输入一个节点，输出这个节点为根的树的深度

```python
def max_depth(root):
    
    if root is None:
        return 0

    return 1+max(max_depth(root.left),max_depth(root.right))
```

```python
assert max_depth(None) == 0
assert max_depth(TreeNode(1)) == 1
assert max_depth(sample_tree()) == 3
assert max_depth(TreeNode(1, None, TreeNode(2, None, TreeNode(3)))) == 3
```

## 4. 相同的树

判断两棵树是否结构相同且对应节点值相等。列出两棵树都空、只有一棵空、都非空三种情况。

契约：输出两个节点，输出是判断这两个节点为根的树是否相同

```python
def is_same_tree(p, q):
    if p is None and q is None:
        return True
    if p is None or q is None:
        return False
    if p.val !=q.val:
        return False
    #最后只剩下两个节点都非空且数值相等，就要继续查看这两个节点的子节点

    if is_same_tree(p.left, q.left) and is_same_tree(p.right, q.right):
        return True
    else:
        return False
```

```python
assert is_same_tree(None, None) is True
assert is_same_tree(TreeNode(1), None) is False
assert is_same_tree(sample_tree(), sample_tree()) is True
assert is_same_tree(TreeNode(1, TreeNode(2)), TreeNode(1, None, TreeNode(2))) is False
```

## 5. 翻转二叉树

交换每个节点的左右子树，返回根节点。允许原地修改。

契约：输入是一个根节点，输出是一个反转后的根节点，副作用是把这个节点下的left和right都交换了

```python
def invert_tree(root):
    # 函数输入：根节点
    # 函数输出：交换后的根节点
    # 空节点情况：返回None
    if root is None:
        return None
    root.left,root.right = root.right,root.left
    invert_tree(root.left)
    invert_tree(root.right)
    return root
```

```python
assert invert_tree(None) is None
root = invert_tree(sample_tree())
assert preorder(root) == [1,3,2,5,4]
```

判断是否存在从根到叶子的路径，其节点值之和等于 `target_sum`。中间节点达到目标不算成功。

```python
# 这个版本是错误的，路径终点判断错误，终点是叶子（left和right都为None）

def has_path_sum(root, target_sum):
    # 函数输入：根节点，目标值
    # 函数输出：布尔值
    # 空节点情况：判断此时remain是否为0，如果为0就是找到了，如果不为0就没找到

    if root is None:
        if target_sum == 0:
            return True
        else:
            return False
        
    remain = target_sum-root.val
    left_result = has_path_sum(root.left,remain)
    right_result = has_path_sum(root.right,remain)

    return left_result or right_result


    
```

契约：输入是根节点和目标值，输出是这棵树中是否有根节点到叶子的路径是目标值

```python
def has_path_sum(root, target_sum):
    # 函数输入：根节点，目标值
    # 函数输出：布尔值
    # 叶子节点情况：判断此时remain是否为叶子节点的值，如果是就是找到了，如果不是就没找到
    # 空节点情况：返回False

    # 关键是要想明白函数的意义

    if root is None:
        return False

    if root.left is None and root.right is None:
        if target_sum == root.val:
            return True
        else:
            return False
        
    remain = target_sum-root.val
    left_result = has_path_sum(root.left,remain)
    right_result = has_path_sum(root.right,remain)

    return left_result or right_result
    
```

```python
#       5
#      / \
#     4   8
#    /     \
#   11      13
#  /  \
# 7    2
root = TreeNode(5, TreeNode(4, TreeNode(11, TreeNode(7), TreeNode(2))), TreeNode(8, None, TreeNode(13)))
assert has_path_sum(root, 22) is True
assert has_path_sum(root, 5) is False
assert has_path_sum(None, 0) is False
assert has_path_sum(TreeNode(-2, None, TreeNode(-3)), -5) is True
```

## 7. 挑战：二叉树直径

返回任意两个节点之间最长路径的边数。路径不一定经过根。

提示：辅助函数向父节点返回什么？经过当前节点的路径怎样由左右结果组成？

```python
#这个版本是错误的，需要知道最长路径不一定会经过root，因此对于每个节点都要统计一下

def diameter_of_binary_tree(root):
    # 输入：根节点
    # 输出：任意两个节点之间最长路径的边数
    
    def height(root):
        # 输入：根节点
        # 输出：这棵树的高度
        # 空节点：return -1

        if root is None:
            return -1 

        return 1+max(height(root.left),height(root.right))

    d=height(root.left)+height(root.right)+2

    return d
```

height(root)契约：输入是根节点，输出是这棵树的边高度，副作用是同时更新了目前遇到的树的最大直径

```python
def diameter_of_binary_tree(root):
    # 输入：根节点
    # 输出：任意两个节点之间最长路径的边数
    # 需要一个辅助变量best，记录最长的边数，最后返回best

    best = 0
    
    def height(root):
        # 输入：根节点
        # 输出：这棵树的高度
        # 空节点：return -1
        nonlocal best

        if root is None:
            return -1 

        leftheight = height(root.left)
        rightheight = height(root.right)
        current_d = leftheight+rightheight+2
        best = max(best,current_d)

        return 1+max(leftheight,rightheight)

    height(root)

    return best
```

```python
assert diameter_of_binary_tree(None) == 0
assert diameter_of_binary_tree(TreeNode(1)) == 0
assert diameter_of_binary_tree(sample_tree()) == 3
assert diameter_of_binary_tree(TreeNode(1, TreeNode(2, TreeNode(3)))) == 2
```

## 8. 哈希复习：异位词判断

重新实现异位词判断：禁止 `sorted()`、`sort()` 和 `Counter`，要求 `O(n)` 时间。

```python
# 使用哈希表，key是每个可能出现的字母，value是出现的次数，对于s是+，对于t是-，最后字典中的value都是0.
# 需要先判断字母的个数相同

def is_anagram_no_sort(s, t):
    if len(s) !=len(t):
        return False
    
    
    dict1 = {}
    for x in s:
        if x in dict1:
            dict1[x] +=1
        else:
            dict1[x] = 1
    
    for y in t:
        if y not in dict1:
            return False
        else:
            dict1[y]-=1
            if dict1[y] < 0: return False
    
    return True

    
```

```python
assert is_anagram_no_sort('', '') is True
assert is_anagram_no_sort('anagram', 'nagaram') is True
assert is_anagram_no_sort('aacc', 'ccac') is False
assert is_anagram_no_sort('ab', 'a') is False
```

## 9. 数组复习：严格常量空间移动零

原地移动所有零到末尾，保持非零元素顺序。禁止切片赋值以及创建长度随 `n` 增长的新列表。返回 `None`。

双指针题目，一个read一个write，原地修改

```python
def move_zeroes_constant_space(nums):
    write = 0
    for read in range(len(nums)):
        if nums[read]  != 0:
            nums[write] = nums[read]
            write+=1

    for x in range(write,len(nums)):
        nums[x] = 0
```

```python
for original, expected in [([0,1,0,3,12],[1,3,12,0,0]), ([],[]), ([0,0],[0,0]), ([1,2],[1,2])]:
    nums = original.copy()
    result = move_zeroes_constant_space(nums)
    assert nums == expected
    assert result is None
```
