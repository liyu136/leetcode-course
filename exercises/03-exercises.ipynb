## 2. 频率统计

实现 `frequency_map(items)`，返回每个元素的出现次数。不要使用 `Counter`。

编码前写出：键表示什么、值表示什么、每轮如何更新。

统计出现次数，需要使用dict   
key表示元素，value表示次数  

```python
def frequency_map(items):
    groups = {}
    for item in items:
        if item not in groups:
            groups[item] = 1
        else:
            groups[item] += 1
    return groups
```

```python
assert frequency_map([]) == {}
assert frequency_map([1, 1, 2, 3, 3, 3]) == {1: 2, 2: 1, 3: 3}
assert frequency_map('banana') == {'b': 1, 'a': 3, 'n': 2}
```

## 3. 有效的字母异位词

如果两个字符串包含完全相同的字符及次数，返回 `True`，否则返回 `False`。不要排序，也不要使用 `Counter`。目标时间 `O(n)`。

示例：`anagram` 与 `nagaram` 为真，`rat` 与 `car` 为假。

```python
def is_anagram(s, t):
    s_sort = "".join(sorted(s))
    t_sort = "".join(sorted(t))
    if s_sort == t_sort:
        return True
    else:
        return False

# 使用了sort，所以时间复杂度是O(nlogn)
# 题目要求时间复杂度是O(n)
```

```python
def is_anagram(s, t):
    if len(s) != len(t):
        return False
    # 确保后边的两个字符串长度相同
    
    count = {}
    for x in s:
        if x not in count:
            count[x]=1
        else:
            count[x]+=1
    
    for y in t:
        if y not in count:
            return False
        else:
            count[y]-=1
            if count[y] < 0:
                return False

    return True
```

```python
assert is_anagram('', '') is True
assert is_anagram('anagram', 'nagaram') is True
assert is_anagram('rat', 'car') is False
assert is_anagram('aacc', 'ccac') is False
```

## 4. 两数之和：`O(n)`

返回两个不同元素的下标，假设恰好有一组答案。不得使用双重循环。

编码前回答：字典的键和值分别是什么？为什么要先查询补数，再写入当前元素？

只能遍历一次  
key存放具体的数字，value存放下标   


```python
def two_sum(nums, target):
    groups = {}
    for i in range(len(nums)):
        needed = target-nums[i]

        if needed in groups:
            return [groups[needed],i]
        groups[nums[i]] = i
```

```python
assert two_sum([2, 7, 11, 15], 9) == [0, 1]
assert two_sum([3, 2, 4], 6) == [1, 2]
assert two_sum([3, 3], 6) == [0, 1]
assert two_sum([-3, 4, 3, 90], 0) == [0, 2]
```

## 5. 两个数组的交集

返回两个数组中共同出现的不同元素。输出顺序不作要求，但每个元素只能出现一次。

先分别写出：使用集合运算的版本；手写遍历和成员查询的版本。

```python
def intersection_builtin(nums1, nums2):
    groups1 = set(nums1)
    groups2 = set(nums2)
    
    comm = groups1 & groups2
    return list(comm)

def intersection_manual(nums1, nums2):
    seen = set(nums1)
    comm = []
    for x in nums2:
        if x in seen:
            comm.append(x)
            seen.remove(x) #这个很重要！
    
    return comm
```

```python
for func in (intersection_builtin, intersection_manual):
    assert set(func([1, 2, 2, 1], [2, 2])) == {2}
    assert set(func([4, 9, 5], [9, 4, 9, 8, 4])) == {4, 9}
    assert func([], [1, 2]) == []
```

## 6. 分组字母异位词

把由相同字符组成的单词放进同一组。组的顺序和组内顺序不作要求。

示例：`['eat','tea','tan','ate','nat','bat']` 应分成 `{eat,tea,ate}`、`{tan,nat}`、`{bat}`。

编码前回答：什么对象可以作为每组的规范键？字典的值是什么？

对每个word排序，key是排序后的word，value是原word的list



```python
def group_anagrams(words):

    groups={}
    for word in words:
        word1 = "".join(sorted(word))
        if word1 not in groups:
            groups[word1] = [word]
        else:
            groups[word1].append(word)
    
    return list(groups.values())
```

```python
def normalize_groups(groups):
    return sorted(sorted(group) for group in groups)

assert normalize_groups(group_anagrams([])) == []
assert normalize_groups(group_anagrams([''])) == [['']]
assert normalize_groups(group_anagrams(['eat','tea','tan','ate','nat','bat'])) == normalize_groups([['eat','tea','ate'], ['tan','nat'], ['bat']])
```

## 7. 挑战：最长连续序列

给定无序整数数组，返回最长连续整数序列的长度。要求平均 `O(n)` 时间，不允许排序。

示例：`[100,4,200,1,3,2] → 4`，因为最长序列是 `1,2,3,4`。

提示：不要从每个数字都向后搜索。先判断什么样的数字才是序列起点。

思路：只搜一遍，对于元素x，如果开始计数，要求x-1不存在



```python
def longest_consecutive(nums):
    seen = set(nums)
    best = 0
    for x in nums: #这里不对，应该扫描seen
        current = x
        if current-1 not in seen:
            length = 1
            i = 1
            while current+i in seen:
                length += 1
                i += 1
            best = max(best,length)
    return best
```

```python
assert longest_consecutive([]) == 0
assert longest_consecutive([100, 4, 200, 1, 3, 2]) == 4
assert longest_consecutive([0, 3, 7, 2, 5, 8, 4, 6, 0, 1]) == 9
assert longest_consecutive([1, 2, 0, 1]) == 3
```
