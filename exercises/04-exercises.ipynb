## 2. 用类实现栈

补全 `ArrayStack`。空栈执行 `pop()` 或 `peek()` 时抛出 `IndexError`。所有操作应为 `O(1)` 或摊还 `O(1)`。

````python
class ArrayStack:
    def __init__(self):
        self.data = []

    def push(self, value):
        self.data.append(value)

    def pop(self):
        if self.data:
            return self.data.pop()
        else:
            raise IndexError("pop from empty stack")

    def peek(self):
        if self.data:
            return self.data[-1]
        else:
            raise IndexError("peek from empty stack")

    def is_empty(self):
        return self.data == []
    # 更好的写法是：return not self.data

    def __len__(self):
        return len(self.data)
````

````python
stack = ArrayStack()
assert stack.is_empty() is True
assert len(stack) == 0
stack.push(10)
stack.push(20)
assert stack.peek() == 20
assert len(stack) == 2
assert stack.pop() == 20
assert stack.pop() == 10
assert stack.is_empty() is True

for method_name in ('pop', 'peek'):
    try:
        getattr(stack, method_name)()
        raise AssertionError('空栈应该抛出 IndexError')
    except IndexError:
        pass
````

## 3. 有效括号

字符串只包含 `()[]{}`。判断括号是否有效。

有效条件：类型正确、顺序正确、全部匹配。先写一句循环不变量。

循环不变量：stack中只存放还没有被匹配上的左括号   
碰到左括号就入栈   
栈顶左括号如果碰到了对应的右括号出栈   

````python
def is_valid_parentheses(text):
    pair={"(":")","[":"]","{":"}"}
    stack = []

    for ch in text:
        if ch in "([{":
            stack.append(ch)
        else:
            if stack and ch == pair[stack[-1]]:
                stack.pop()
            else:
                return False
            
    return not stack
````

````python
assert is_valid_parentheses('') is True
assert is_valid_parentheses('()[]{}') is True
assert is_valid_parentheses('([{}])') is True
assert is_valid_parentheses('(]') is False
assert is_valid_parentheses('([)]') is False
assert is_valid_parentheses('(((') is False
assert is_valid_parentheses(']') is False
````

## 4. 删除相邻重复字符

反复删除相邻且相同的两个字符，直到不能再删。返回最终字符串。

示例：`abbaca → ca`。目标时间 `O(n)`。

栈：存放邻近没有相同的字符串   
如果和栈顶不相同可以入栈  
如果和栈顶相同，栈顶出栈   



````python
def remove_adjacent_duplicates(text):
    stack= []

    for ch in text:
        if stack and ch == stack[-1]:
            stack.pop()
        else: 
            stack.append(ch)
    return "".join(stack)
````

````python
assert remove_adjacent_duplicates('') == ''
assert remove_adjacent_duplicates('abbaca') == 'ca'
assert remove_adjacent_duplicates('azxxzy') == 'ay'
assert remove_adjacent_duplicates('aaaa') == ''
assert remove_adjacent_duplicates('abc') == 'abc'
````

## 5. 逆波兰表达式求值

运算符包括 `+ - * /`，除法向零截断。输入保证表达式有效。

编码前说明：栈中保存什么？遇到运算符时先弹出的数是左操作数还是右操作数？

栈里边只保存待运算的数字  
遇到计算符号开始出栈并计算，每次弹两个  
计算结果会再次入栈   
先弹出来的是右操作数  




````python
def evaluate_rpn(tokens):
    stack = []
    for ch in tokens:

        if ch == "+":
            right = stack.pop()
            left = stack.pop()
            stack.append(right+left)
        elif ch == "-":   
            right = stack.pop()
            left = stack.pop()
            stack.append(left-right)
        elif ch == "*":   
            right = stack.pop()
            left = stack.pop()
            stack.append(left*right)
        elif ch == "/":   
            right = stack.pop()
            left = stack.pop()
            stack.append(int(left/right))
            #不产生浮点数的写法：
            # value = abs(left) // abs(right)
            # if (left < 0) != (right < 0):
            #     value = -value
            # stack.append(value)

        else:
            stack.append(int(ch))
    return stack[0]
````

````python
assert evaluate_rpn(['2', '1', '+', '3', '*']) == 9
assert evaluate_rpn(['4', '13', '5', '/', '+']) == 6
assert evaluate_rpn(['7', '-3', '/']) == -2
assert evaluate_rpn(['3', '4', '-']) == -1
````

## 6. MinStack 设计

实现支持 `push`、`pop`、`top`、`get_min` 的栈，所有操作必须为 `O(1)`。输入测试不会在空栈上调用后三个方法。

禁止在 `get_min` 中调用 `min(self.values)`。先写出额外状态表示什么。

````python
class MinStack:
    def __init__(self):
        self.data = []
        self.min = []

    def push(self, value):
        self.data.append(value)
        if self.min and value < self.min[-1]:
            self.min.append(value)
        elif not self.min:
            self.min.append(value)
        else:
            self.min.append(self.min[-1])


    def pop(self):
        self.min.pop()
        return self.data.pop()
    

    def top(self):
        return self.data[-1]

    def get_min(self):
        return self.min[-1]
        
````

````python
stack = MinStack()
stack.push(-2)
stack.push(0)
stack.push(-3)
assert stack.get_min() == -3
assert stack.pop() == -3
assert stack.top() == 0
assert stack.get_min() == -2
stack.push(-2)
assert stack.get_min() == -2
````

## 7. 挑战：每日温度

对每一天，返回还要等待几天才会出现更高温度；如果以后没有更高温度，答案为 0。要求 `O(n)` 时间。

提示：栈中保存尚未找到更高温度的下标。写出从栈底到栈顶对应温度满足什么单调性质。

栈中保存的是递减温度的天数index，也是没有还没有计算出answer的对应的天数。

````python
def daily_temperatures(temperatures):
    answer = [0]*len(temperatures)
    day = [] #stack

    for i,temper in enumerate(temperatures):
        while day and temper > temperatures[day[-1]]:
            answer[day[-1]] = i-day[-1]
            day.pop()  

        day.append(i)
    return answer
````

````python
assert daily_temperatures([]) == []
assert daily_temperatures([73,74,75,71,69,72,76,73]) == [1,1,4,2,1,1,0,0]
assert daily_temperatures([30,40,50,60]) == [1,1,1,0]
assert daily_temperatures([30,60,90]) == [1,1,0]
assert daily_temperatures([90,80,70]) == [0,0,0]
````
